*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
def check_game_status(board):
    """
    Returns the status of a tic-tac-toe game.
    
    Args:
        board: n×n matrix where 0=empty, 1=X, 2=O
    
    Returns:
        -1: Game incomplete (has empty spaces, no winner)
         0: Draw (full board, no winner)
         1: Player 1 (X) won
         2: Player 2 (O) won
    """
    n = len(board)
    
    def check_winner(player):
        # Check rows
        for i in range(n):
            if all(board[i][j] == player for j in range(n)):
                return True
        
        # Check columns
        for j in range(n):
            if all(board[i][j] == player for i in range(n)):
                return True
        
        # Check main diagonal
        if all(board[i][i] == player for i in range(n)):
            return True
        
        # Check anti-diagonal
        if all(board[i][n - 1 - i] == player for i in range(n)):
            return True
        
        return False
    
    if check_winner(1):
        return 1
    if check_winner(2):
        return 2
    
    # Check for empty spaces
    has_empty = any(board[i][j] == 0 for i in range(n) for j in range(n))
    return -1 if has_empty else 0
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
def place_mark(board, player, row, col):
    """
    Places a mark on the tic-tac-toe board.
    
    Args:
        board: n×n matrix where 0=empty, 1=X, 2=O
        player: 1 for X, 2 for O
        row: row index
        col: column index
    
    Returns:
        True if move was successful, False otherwise
    """
    n = len(board)
    
    # Check if coordinates are within bounds
    if row < 0 or row >= n or col < 0 or col >= n:
        return False
    
    # Check if location is empty
    if board[row][col] != 0:
        return False
    
    # Place the mark
    board[row][col] = player
    return True


# QUICK TEST
if __name__ == "__main__":
    board = [[0, 0, 0], [0, 0, 0], [0, 0, 0]]
    
    # Valid move
    print(place_mark(board, 1, 0, 0))  # True
    print(board)  # [[1, 0, 0], [0, 0, 0], [0, 0, 0]]
    
    # Invalid move - already occupied
    print(place_mark(board, 2, 0, 0))  # False
    print(board)  # Unchanged
    
    # Valid move - player 2
    print(place_mark(board, 2, 1, 1))  # True
    print(board)  # [[1, 0, 0], [0, 2, 0], [0, 0, 0]]
    
    # Invalid move - out of bounds
    print(place_mark(board, 1, 5, 5))  # False
````

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,
- repeatedly ask two players for a location,
- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
def create_board(n):
    """Creates an empty n×n tic-tac-toe board."""
    return [[0 for _ in range(n)] for _ in range(n)]


def draw_board_with_labels(board):
    """Draws a tic-tac-toe board with row and column labels."""
    symbols = {0: " ", 1: "X", 2: "O"}
    n = len(board)
    
    # Print column headers (A, B, C, ...)
    col_labels = " ".join([chr(65 + i) for i in range(n)])
    print("  " + col_labels)
    
    # Print rows with row numbers
    for i in range(n):
        cell_symbols = [symbols[value] for value in board[i]]
        row_line = "|".join(cell_symbols)
        print(f"{i + 1} {row_line}")
        
        if i < n - 1:
            print("  " + "-" * (len(row_line)))


def parse_position(position):
    """Converts a position like "A1" to (row, col) coordinates."""
    if len(position) != 2:
        return None
    
    col_char = position[0].upper()
    row_char = position[1]
    
    if not col_char.isalpha():
        return None
    col = ord(col_char) - ord('A')
    
    if not row_char.isdigit():
        return None
    row = int(row_char) - 1
    
    return (row, col)


def place_mark(board, player, row, col):
    """Places a mark on the board. Returns True if successful, False otherwise."""
    n = len(board)
    
    if row < 0 or row >= n or col < 0 or col >= n:
        return False
    
    if board[row][col] != 0:
        return False
    
    board[row][col] = player
    return True


def make_move(board, player, position):
    """Makes a move using a position string like "A1"."""
    coords = parse_position(position)
    
    if coords is None:
        return False
    
    row, col = coords
    return place_mark(board, player, row, col)


def get_player_move(board, player):
    """Gets input from the player and makes a move."""
    symbol = "X" if player == 1 else "O"
    
    while True:
        position = input(f"Player {symbol}, enter your move (e.g., A1, B2, C3): ").strip()
        
        if make_move(board, player, position):
            break
        else:
            print("Invalid move! Cell already occupied or out of bounds. Try again.")


def check_game_status(board):
    """
    Returns the status of a tic-tac-toe game.
    
    Returns:
        -1: Game incomplete
         0: Draw
         1: Player 1 (X) won
         2: Player 2 (O) won
    """
    n = len(board)
    
    def check_winner(player):
        # Check rows
        for i in range(n):
            if all(board[i][j] == player for j in range(n)):
                return True
        
        # Check columns
        for j in range(n):
            if all(board[i][j] == player for i in range(n)):
                return True
        
        # Check main diagonal
        if all(board[i][i] == player for i in range(n)):
            return True
        
        # Check anti-diagonal
        if all(board[i][n - 1 - i] == player for i in range(n)):
            return True
        
        return False
    
    if check_winner(1):
        return 1
    if check_winner(2):
        return 2
    
    has_empty = any(board[i][j] == 0 for i in range(n) for j in range(n))
    return -1 if has_empty else 0


def play_tic_tac_toe():
    """Runs a full tic-tac-toe game."""
    n = 3
    board = create_board(n)
    current_player = 1
    
    print("=" * 30)
    print("Welcome to Tic-Tac-Toe!")
    print("=" * 30)
    print()
    
    while True:
        # Draw board
        draw_board_with_labels(board)
        print()
        
        # Get player move
        get_player_move(board, current_player)
        print()
        
        # Check game status
        status = check_game_status(board)
        
        if status == 1:
            draw_board_with_labels(board)
            print("🎉 Player X wins!")
            break
        elif status == 2:
            draw_board_with_labels(board)
            print("🎉 Player O wins!")
            break
        elif status == 0:
            draw_board_with_labels(board)
            print("🤝 It's a draw!")
            break
        
        # Switch player
        current_player = 2 if current_player == 1 else 1


if __name__ == "__main__":
    play_tic_tac_toe()
````
