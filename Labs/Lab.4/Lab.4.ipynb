1. Write a "counter" class that can be incremented up to a specified maximum value, will print an error if an attempt is made to increment beyond that value, and allows reseting the counter. 

```python
class Counter:
    def __init__(self, max_value):
        self.count = 0
        self.max_value = max_value

    def increment(self):
        if self.count < self.max_value:
            self.count += 1
        else:
            print("Error: Cannot increment past the maximum value of", self.max_value)

    def reset(self):
        self.count = 0
        print("Counter has been reset to 0.")
```

```python
c=Counter(4)
c.increment()
print(c.increment())
print(c.count)
```

6. Implement a triangle class analogous to the rectangle and circle in question 5.

```python
import math

class Triangle(Shape):
    def __init__(self, side_a, side_b, side_c, x, y):
        super().__init__(x, y)
        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c

    def get_side_a(self):
        return self.__side_a

    def get_side_b(self):
        return self.__side_b

    def get_side_c(self):
        return self.__side_c

    def area(self):        
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        s = (a + b + c) / 2   # semi-perimeter
        return math.sqrt(s * (s - a) * (s - b) * (s - c))

    def perimeter(self):
        return self.__side_a + self.__side_b + self.__side_c
```

7. Add a function to the object classes, including the base, that returns a list of up to 16 pairs of  $x$ and $y$ points on the perimeter of the object. 

```python
import math

class Shape:
    def __init__(self, x, y):
        self._x = x   
        self._y = y   

    def get_x(self):
        return self._x

    def get_y(self):
        return self._y

    def area(self):
        raise NotImplementedError("Subclass must implement area()")

    def perimeter(self):
        raise NotImplementedError("Subclass must implement perimeter()")

    def perimeter_points(self):
        raise NotImplementedError("Subclass must implement perimeter_points()")

    def _points_on_polygon(self, vertices, num_points=16):
        n_edges = len(vertices)
        base = num_points // n_edges
        extra = num_points % n_edges   

        points = []
        for i in range(n_edges):
            x0, y0 = vertices[i]
            x1, y1 = vertices[(i + 1) % n_edges]

            n_this_edge = base + (1 if i < extra else 0)

            for k in range(n_this_edge):
                t = k / n_this_edge   
                px = x0 + t * (x1 - x0)
                py = y0 + t * (y1 - y0)
                points.append((px, py))

        return points


class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length = length
        self.__width = width

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def perimeter_points(self):
      
        corners = [
            (self._x, self._y),
            (self._x + self.__length, self._y),
            (self._x + self.__length, self._y + self.__width),
            (self._x, self._y + self.__width),
        ]
        return self._points_on_polygon(corners, num_points=16)


class Circle(Shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius

    def get_radius(self):
        return self.__radius

    def area(self):
        return math.pi * self.__radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.__radius

    def perimeter_points(self):
        points = []
        num_points = 16
        for i in range(num_points):
            angle = 2 * math.pi * i / num_points   
            px = self._x + self.__radius * math.cos(angle)
            py = self._y + self.__radius * math.sin(angle)
            points.append((px, py))
        return points


class Triangle(Shape):
    def __init__(self, side_a, side_b, side_c, x, y):
        super().__init__(x, y)
        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c

    def get_side_a(self):
        return self.__side_a

    def get_side_b(self):
        return self.__side_b

    def get_side_c(self):
        return self.__side_c

    def area(self):
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        s = (a + b + c) / 2
        return math.sqrt(s * (s - a) * (s - b) * (s - c))

    def perimeter(self):
        return self.__side_a + self.__side_b + self.__side_c

    def _vertices(self):
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        vertex_a = (self._x, self._y)
        vertex_b = (self._x + c, self._y)

        angle_a = math.acos((b**2 + c**2 - a**2) / (2 * b * c))
        vertex_c = (self._x + b * math.cos(angle_a), self._y + b * math.sin(angle_a))

        return [vertex_a, vertex_b, vertex_c]

    def perimeter_points(self):
        return self._points_on_polygon(self._vertices(), num_points=16)
```

```python
shapes = [Rectangle(5, 3, 0, 0), Circle(4, 0, 0), Triangle(3, 4, 5, 0, 0)]
for s in shapes:
    print(type(s).__name__)
    print("perimeter points:", s.perimeter_points())
    print()
```

8. Add a function to the object classes, including the base, that tests if a given set of $x$ and $y$ coordinates are inside of the object. You'll have to think through how to determine if a set of coordinates are inside an object for each object type.

```python
import math

class Shape:
    def __init__(self, x, y):
        self._x = x   
        self._y = y   
        
    def get_x(self):
        return self._x

    def get_y(self):
        return self._y

    def area(self):
        raise NotImplementedError("Subclass must implement area()")

    def perimeter(self):
        raise NotImplementedError("Subclass must implement perimeter()")

    def perimeter_points(self):
        raise NotImplementedError("Subclass must implement perimeter_points()")

    def contains_point(self, px, py):
        raise NotImplementedError("Subclass must implement contains_point()")

    def _points_on_polygon(self, vertices, num_points=16):
        n_edges = len(vertices)
        base = num_points // n_edges
        extra = num_points % n_edges

        points = []
        for i in range(n_edges):
            x0, y0 = vertices[i]
            x1, y1 = vertices[(i + 1) % n_edges]
            n_this_edge = base + (1 if i < extra else 0)
            for k in range(n_this_edge):
                t = k / n_this_edge
                px = x0 + t * (x1 - x0)
                py = y0 + t * (y1 - y0)
                points.append((px, py))
        return points


class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length = length
        self.__width = width

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def perimeter_points(self):
        corners = [
            (self._x, self._y),
            (self._x + self.__length, self._y),
            (self._x + self.__length, self._y + self.__width),
            (self._x, self._y + self.__width),
        ]
        return self._points_on_polygon(corners, num_points=16)

    def contains_point(self, px, py):
        within_x = self._x <= px <= self._x + self.__length
        within_y = self._y <= py <= self._y + self.__width
        return within_x and within_y


class Circle(Shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius

    def get_radius(self):
        return self.__radius

    def area(self):
        return math.pi * self.__radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.__radius

    def perimeter_points(self):
        points = []
        num_points = 16
        for i in range(num_points):
            angle = 2 * math.pi * i / num_points
            px = self._x + self.__radius * math.cos(angle)
            py = self._y + self.__radius * math.sin(angle)
            points.append((px, py))
        return points

    def contains_point(self, px, py):
        distance = math.sqrt((px - self._x) ** 2 + (py - self._y) ** 2)
        return distance <= self.__radius


class Triangle(Shape):
    def __init__(self, side_a, side_b, side_c, x, y):
        super().__init__(x, y)
        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c

    def get_side_a(self):
        return self.__side_a

    def get_side_b(self):
        return self.__side_b

    def get_side_c(self):
        return self.__side_c

    def area(self):
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        s = (a + b + c) / 2
        return math.sqrt(s * (s - a) * (s - b) * (s - c))

    def perimeter(self):
        return self.__side_a + self.__side_b + self.__side_c

    def _vertices(self):
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        vertex_a = (self._x, self._y)
        vertex_b = (self._x + c, self._y)
        angle_a = math.acos((b**2 + c**2 - a**2) / (2 * b * c))
        vertex_c = (self._x + b * math.cos(angle_a), self._y + b * math.sin(angle_a))
        return [vertex_a, vertex_b, vertex_c]

    def perimeter_points(self):
        return self._points_on_polygon(self._vertices(), num_points=16)

    @staticmethod
    def _sign(p1, p2, p3):
        return (p1[0] - p3[0]) * (p2[1] - p3[1]) - (p2[0] - p3[0]) * (p1[1] - p3[1])

    def contains_point(self, px, py):
        v1, v2, v3 = self._vertices()
        pt = (px, py)

        d1 = self._sign(pt, v1, v2)
        d2 = self._sign(pt, v2, v3)
        d3 = self._sign(pt, v3, v1)

        has_negative = (d1 < 0) or (d2 < 0) or (d3 < 0)
        has_positive = (d1 > 0) or (d2 > 0) or (d3 > 0)
        return not (has_negative and has_positive)
```

```python
import math
class Shape:
    def __init__(self, x, y):
        self._x = x   
        self._y = y  

    def get_x(self):
        return self._x

    def get_y(self):
        return self._y

    def area(self):
        raise NotImplementedError("Subclass must implement area()")

    def perimeter(self):
        raise NotImplementedError("Subclass must implement perimeter()")

    def perimeter_points(self):
        raise NotImplementedError("Subclass must implement perimeter_points()")

    def contains_point(self, px, py):
        raise NotImplementedError("Subclass must implement contains_point()")

    def overlaps(self, other):
        
        if self.contains_point(other.get_x(), other.get_y()):
            return True
        if other.contains_point(self._x, self._y):
            return True

        
        for (px, py) in self.perimeter_points():
            if other.contains_point(px, py):
                return True

        for (px, py) in other.perimeter_points():
            if self.contains_point(px, py):
                return True

        return False

    def _points_on_polygon(self, vertices, num_points=16):
        n_edges = len(vertices)
        base = num_points // n_edges
        extra = num_points % n_edges

        points = []
        for i in range(n_edges):
            x0, y0 = vertices[i]
            x1, y1 = vertices[(i + 1) % n_edges]
            n_this_edge = base + (1 if i < extra else 0)
            for k in range(n_this_edge):
                t = k / n_this_edge
                px = x0 + t * (x1 - x0)
                py = y0 + t * (y1 - y0)
                points.append((px, py))
        return points


class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length = length
        self.__width = width

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def perimeter_points(self):
        corners = [
            (self._x, self._y),
            (self._x + self.__length, self._y),
            (self._x + self.__length, self._y + self.__width),
            (self._x, self._y + self.__width),
        ]
        return self._points_on_polygon(corners, num_points=16)

    def contains_point(self, px, py):
        within_x = self._x <= px <= self._x + self.__length
        within_y = self._y <= py <= self._y + self.__width
        return within_x and within_y


class Circle(Shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius

    def get_radius(self):
        return self.__radius

    def area(self):
        return math.pi * self.__radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.__radius

    def perimeter_points(self):
        points = []
        num_points = 16
        for i in range(num_points):
            angle = 2 * math.pi * i / num_points
            px = self._x + self.__radius * math.cos(angle)
            py = self._y + self.__radius * math.sin(angle)
            points.append((px, py))
        return points

    def contains_point(self, px, py):
        distance = math.sqrt((px - self._x) ** 2 + (py - self._y) ** 2)
        return distance <= self.__radius


class Triangle(Shape):
    def __init__(self, side_a, side_b, side_c, x, y):
        super().__init__(x, y)
        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c

    def get_side_a(self):
        return self.__side_a

    def get_side_b(self):
        return self.__side_b

    def get_side_c(self):
        return self.__side_c

    def area(self):
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        s = (a + b + c) / 2
        return math.sqrt(s * (s - a) * (s - b) * (s - c))

    def perimeter(self):
        return self.__side_a + self.__side_b + self.__side_c

    def _vertices(self):
        a, b, c = self.__side_a, self.__side_b, self.__side_c
        vertex_a = (self._x, self._y)
        vertex_b = (self._x + c, self._y)
        angle_a = math.acos((b**2 + c**2 - a**2) / (2 * b * c))
        vertex_c = (self._x + b * math.cos(angle_a), self._y + b * math.sin(angle_a))
        return [vertex_a, vertex_b, vertex_c]

    def perimeter_points(self):
        return self._points_on_polygon(self._vertices(), num_points=16)

    @staticmethod
    def _sign(p1, p2, p3):
        return (p1[0] - p3[0]) * (p2[1] - p3[1]) - (p2[0] - p3[0]) * (p1[1] - p3[1])

    def contains_point(self, px, py):
        v1, v2, v3 = self._vertices()
        pt = (px, py)
        d1 = self._sign(pt, v1, v2)
        d2 = self._sign(pt, v2, v3)
        d3 = self._sign(pt, v3, v1)
        has_negative = (d1 < 0) or (d2 < 0) or (d3 < 0)
        has_positive = (d1 > 0) or (d2 > 0) or (d3 > 0)
        return not (has_negative and has_positive)

```

11. Create a `RasterDrawing` class. Demonstrate that you can create a drawing made of several shapes, paint the drawing, modify the drawing, and paint it again. 

```python
from paint import Canvas, Rectangle, Circle, Triangle, CompoundShape


class RasterDrawing:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.shapes = []

    def add_shape(self, shape):
        self.shapes.append(shape)

    def remove_shape(self, shape_or_index):
       
        if isinstance(shape_or_index, int):
            self.shapes.pop(shape_or_index)
        else:
            self.shapes.remove(shape_or_index)

    def paint(self):
        canvas = Canvas(self.width, self.height)
        for shape in self.shapes:
            shape.paint(canvas)
        canvas.display()


if __name__ == "__main__":
    drawing = RasterDrawing(width=40, height=20)

  
    rect = Rectangle(length=8, width=10, x=1, y=1)
    circle = Circle(radius=5, x=25, y=8)
    triangle = Triangle(side_a=8, side_b=8, side_c=8, x=15, y=2)

    drawing.add_shape(rect)
    drawing.add_shape(circle)
    drawing.add_shape(triangle)

    print("Original drawing:")
    drawing.paint()

    
    drawing.remove_shape(rect)

    post = Rectangle(length=6, width=1, x=3, y=14)
    head = Circle(radius=3, x=3, y=14)
    balloon = CompoundShape([post, head])
    drawing.add_shape(balloon)
    drawing.add_shape(Rectangle(length=10, width=4, x=20, y=14))

    print("\nModified drawing:")
    drawing.paint()

    print(f"\nThe drawing now has {len(drawing.shapes)} shapes.")
```

```python

import math

class Canvas:
   

    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.grid = [[' ' for _ in range(width)] for _ in range(height)]

    def set_pixel(self, x, y, char):
        col = int(round(x))
        row = int(round(y))
        if 0 <= row < self.height and 0 <= col < self.width:
            self.grid[row][col] = char

    def display(self):
        border = '+' + '-' * self.width + '+'
        print(border)
        for row in self.grid:
            print('|' + ''.join(row) + '|')
        print(border)


def _draw_line(canvas, x1, y1, x2, y2, char):
   
    steps = int(max(abs(x2 - x1), abs(y2 - y1)))
    if steps == 0:
        canvas.set_pixel(x1, y1, char)
        return
    for i in range(steps + 1):
        t = i / steps
        canvas.set_pixel(x1 + (x2 - x1) * t, y1 + (y2 - y1) * t, char)


class Shape:
   

    def area(self):
        raise NotImplementedError

    def perimeter(self):
        raise NotImplementedError

    def contains_point(self, px, py):
        raise NotImplementedError

    def bounding_box(self):
      
        raise NotImplementedError

    def paint(self, canvas):
        raise NotImplementedError

    def overlaps(self, other, resolution=60):
      
        ax1, ay1, ax2, ay2 = self.bounding_box()
        bx1, by1, bx2, by2 = other.bounding_box()

        ix1, iy1 = max(ax1, bx1), max(ay1, by1)
        ix2, iy2 = min(ax2, bx2), min(ay2, by2)

        if ix1 > ix2 or iy1 > iy2:
            return False

        for i in range(resolution + 1):
            for j in range(resolution + 1):
                px = ix1 + (ix2 - ix1) * i / resolution
                py = iy1 + (iy2 - iy1) * j / resolution
                if self.contains_point(px, py) and other.contains_point(px, py):
                    return True
        return False


class Rectangle(Shape):


    def __init__(self, length, width, x, y):
        self.length = length
        self.width = width
        self.x = x
        self.y = y

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

    def bounding_box(self):
        return (self.x, self.y, self.x + self.length, self.y + self.width)

    def contains_point(self, px, py):
        return (self.x <= px <= self.x + self.length and
                self.y <= py <= self.y + self.width)

    def paint(self, canvas):
        for i in range(self.length):
            canvas.set_pixel(self.x + i, self.y, '#')
            canvas.set_pixel(self.x + i, self.y + self.width - 1, '#')
        for j in range(self.width):
            canvas.set_pixel(self.x, self.y + j, '#')
            canvas.set_pixel(self.x + self.length - 1, self.y + j, '#')

    def __repr__(self):
        return (f"Rectangle(length={self.length!r}, width={self.width!r}, "
                f"x={self.x!r}, y={self.y!r})")


class Circle(Shape):
   

    def __init__(self, radius, x, y):
        self.radius = radius
        self.x = x
        self.y = y

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius

    def bounding_box(self):
        return (self.x - self.radius, self.y - self.radius,
                self.x + self.radius, self.y + self.radius)

    def contains_point(self, px, py):
        return (px - self.x) ** 2 + (py - self.y) ** 2 <= self.radius ** 2

    def paint(self, canvas):
        steps = max(36, int(2 * math.pi * self.radius * 4))
        for i in range(steps):
            theta = 2 * math.pi * i / steps
            px = self.x + self.radius * math.cos(theta)
            py = self.y + self.radius * math.sin(theta)
            canvas.set_pixel(px, py, 'O')

    def __repr__(self):
        return f"Circle(radius={self.radius!r}, x={self.x!r}, y={self.y!r})"


class Triangle(Shape):
 

    def __init__(self, side_a, side_b, side_c, x, y):
        self.side_a = side_a
        self.side_b = side_b
        self.side_c = side_c
        self.x = x
        self.y = y
        self._compute_vertices()

    def _compute_vertices(self):
        a, b, c = self.side_a, self.side_b, self.side_c
        cos_a = (a ** 2 + b ** 2 - c ** 2) / (2 * a * b)
        cos_a = max(-1.0, min(1.0, cos_a))
        angle_a = math.acos(cos_a)

        height = b * math.sin(angle_a)
        apex_offset = b * math.cos(angle_a)

        base_y = self.y + height
        self.vA = (self.x, base_y)
        self.vB = (self.x + a, base_y)
        self.vC = (self.x + apex_offset, self.y)

    def area(self):
        a, b, c = self.side_a, self.side_b, self.side_c
        s = (a + b + c) / 2
        return math.sqrt(max(0.0, s * (s - a) * (s - b) * (s - c)))

    def perimeter(self):
        return self.side_a + self.side_b + self.side_c

    def bounding_box(self):
        xs = [self.vA[0], self.vB[0], self.vC[0]]
        ys = [self.vA[1], self.vB[1], self.vC[1]]
        return (min(xs), min(ys), max(xs), max(ys))

    def contains_point(self, px, py):
        def sign(p1, p2, p3):
            return ((p1[0] - p3[0]) * (p2[1] - p3[1]) -
                     (p2[0] - p3[0]) * (p1[1] - p3[1]))

        p = (px, py)
        d1 = sign(p, self.vA, self.vB)
        d2 = sign(p, self.vB, self.vC)
        d3 = sign(p, self.vC, self.vA)

        has_neg = (d1 < 0) or (d2 < 0) or (d3 < 0)
        has_pos = (d1 > 0) or (d2 > 0) or (d3 > 0)
        return not (has_neg and has_pos)

    def paint(self, canvas):
        _draw_line(canvas, self.vA[0], self.vA[1], self.vB[0], self.vB[1], '^')
        _draw_line(canvas, self.vB[0], self.vB[1], self.vC[0], self.vC[1], '^')
        _draw_line(canvas, self.vC[0], self.vC[1], self.vA[0], self.vA[1], '^')

    def __repr__(self):
        return (f"Triangle(side_a={self.side_a!r}, side_b={self.side_b!r}, "
                f"side_c={self.side_c!r}, x={self.x!r}, y={self.y!r})")


class CompoundShape(Shape):
  

    def __init__(self, shapes):
        self.shapes = list(shapes)

    def area(self):
        return sum(s.area() for s in self.shapes)

    def perimeter(self):
        return sum(s.perimeter() for s in self.shapes)

    def bounding_box(self):
        boxes = [s.bounding_box() for s in self.shapes]
        min_x = min(b[0] for b in boxes)
        min_y = min(b[1] for b in boxes)
        max_x = max(b[2] for b in boxes)
        max_y = max(b[3] for b in boxes)
        return (min_x, min_y, max_x, max_y)

    def contains_point(self, px, py):
        return any(s.contains_point(px, py) for s in self.shapes)

    def paint(self, canvas):
        for s in self.shapes:
            s.paint(canvas)

    def __repr__(self):
        return f"CompoundShape({self.shapes!r})"
```

```python
class foo:
    def __init__(self,a,b=None):
        self.a=a
        self.b=b
        
    def __repr__(self):
        return "foo("+repr(self.a)+","+repr(self.b)+")"
    
    def save(self,filename):
        f=open(filename,"w")
        f.write(self.__repr__())
        f.close()
        
   
def foo_loader(filename):
    f=open(filename,"r")
    tmp=eval(f.read())
    f.close()
    return tmp
```
