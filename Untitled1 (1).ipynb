```python
import uproot
import awkward as ak
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

path = r"C:\Users\USER\Downloads\T_sand.root"

print("File exists:", os.path.exists(path))
print("File size:", round(os.path.getsize(path) / 1024**2, 2), "MB")

f = uproot.open(path)

print("ROOT file opened successfully!")
```

```python
print(f.keys())
```

```python
for name, classname in f.classnames().items():
    print(f"{name:40} {classname}")
```

```python
trees = []

for name, classname in f.classnames().items():
    if "TTree" in classname or "RNTuple" in classname:
        trees.append(name)

print("Trees found:", trees)
```

```python
tree = f[trees[0]]

print("Tree:", trees[0])
print("Number of events:", tree.num_entries)
```

```python
tree.show()
```

```python
tree.show(full_paths=True)
```

```python
import awkward as ak

sample = tree.arrays(
    entry_start=0,
    entry_stop=5,
    library="ak"
)

print(ak.to_list(sample[0]))
```

```python
# Read first 100 events only
sample = tree.arrays(
    entry_start=0,
    entry_stop=100,
    library="ak"
)

# Convert nested ROOT structure to Python records
records = ak.to_list(sample)

# Normalize nested structures into columns
df = pd.json_normalize(records)

print("Shape:", df.shape)
df.head(10)
```
